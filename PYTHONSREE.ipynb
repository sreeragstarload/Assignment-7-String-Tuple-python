```python
#slicing andindex
text = "Hello sree, welcome to Python programming"
#a. print the first character
print("a.",text[0])
#b. print the last character 
print("b.",text[-1])
#c. print the first 5 character 
print("c.",text[:5])
#d. print the last 11 character
print("d.",text[-11:])
#e. print the string in resverse
print("e.",text[::-1])
#f. print the word python
print("f.",text[22:29])
```

```python
strm = "python beginner tutorial"
#a. converted to upper case
print("uppercase:",strm.upper())
#b.converted to lower case
print("lowercase:",strm.lower())
print("capitalized:",strm.capitalize())
print("count of 't'",strm.count("t"))
print("Replace:",strm.replace("python","Machine Learing"))

```

```python
#tuple
t1 = (10,20,30)
t2 = (40,50,60)
t_combine = t1 + t2
print("combined tuple:",t_combine)
print("repeated tuple:",t_combine*3)
print("3rd element:",t_combine[2])
print(" first three element:",t_combine[:3])
print("last three element:",t_combine[3:])
```
